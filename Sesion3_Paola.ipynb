```python
# Dividir la cuenta

cuenta = 500000
propina=15
amigos= 6

# Regresa la cuenta mas la propina

def con_propina(cuenta,porcentaje):

  return cuenta + (cuenta*porcentaje/100)


  #Regresa cuanto paga cada persona


def por_persona(total, personas):
  return total / personas

total = con_propina(cuenta, propina)
pago = por_persona(total, amigos)
print("Total con propina:", total)
print("Cada persona paga:", pago)

```
