**Clientes Inactivos**

```python
import pandas as pd

# DataFrame clientes existente
clientes = pd.DataFrame({
    'cliente_id': ["C001", "C002", "C003", "C004", "C005"],
    'nombre': ["Ana", "Luis", "Sofia", "Pedro", "Lucia"]
})

# DataFrame pedidos existente
pedidos = pd.DataFrame({
    'pedido_id': [101, 102, 103, 104],
    'cliente_id': ["C001", "C003", "C005", "C001"]
})

# 1. Agregar clientes nuevos (C006-C008)
nuevos_clientes = pd.DataFrame({
    'cliente_id': ['C006', 'C007', 'C008'],
    'nombre': ['Paola', 'Cristian', 'Nicolas']
})

clientes = pd.concat([clientes, nuevos_clientes], ignore_index=True)
```

```python
# 2. LEFT JOIN para identificar clientes sin compras

clientes_pedidos = clientes.merge(

pedidos,

on='cliente_id',

how='left'

)

clientes_sin_compras = clientes_pedidos[

clientes_pedidos['pedido_id'].isna()

]

print("Clientes sin compras:")

print(clientes_sin_compras[['cliente_id', 'nombre']])
```

```python
# 3. OUTER JOIN con indicador

merge_resultado = clientes.merge(

pedidos,

on='cliente_id',

how='outer',

indicator=True

)

# Contar categorías del merge
conteo = merge_resultado['_merge'].value_counts()

print("\nConteo por categoría:")
print(conteo)
```

```python
# 4. Lista final para marketing

lista_marketing = clientes_sin_compras[['cliente_id', 'nombre']]

print("\nLista final para marketing:")

print(lista_marketing)
```
