```python
# Ejercicio 1: Clasificador de notas
```

```python
# Crear función
def clasificar_nota(nota):

    if nota < 0 or nota > 5:
        return "Nota inválida"

    elif nota < 3.0:
        return "Reprueba"

    elif nota <= 4.4:
        return "Aprueba"

    else:
        return "Excelente"
```

```python
# Prueba de la función creada

print(clasificar_nota(2.5))
print(clasificar_nota(3.8))
print(clasificar_nota(4.7))
print(clasificar_nota(5.2))
```

```python
# Creación lista de notas

notas = [2.5, 3.8, 4.7, 5.2, 3.0]
```

```python
# Función for

for nota in notas:
    print(nota, "-", clasificar_nota(nota))
```

```python
# Contar los aprobados

aprobados = 0
```

```python
# Contar

for nota in notas:

    resultado = clasificar_nota(nota)

    if resultado == "Aprueba" or resultado == "Excelente":
        aprobados += 1
```

```python
# Resultado final

print("Estudiandtes aprobados:", aprobados)
```

```python
# Ejercicio 2: Carrito de compras
```

```python
# Creación de diccionario

carrito = {
    "camiseta": (45000, 2),
    "jean": (120000, 1),
    "medias": (12000, 3)
}
```

```python
# Verificar que guarde

print(carrito)
```

```python
# Creación de variable

subtotal = 0
```

```python
# Validar carrito y subtotal

for producto, datos in carrito.items():

    precio = datos[0]
    cantidad = datos[1]

    subtotal += precio * cantidad

    print(producto, "-", precio, "x", cantidad, "=",
          precio * cantidad)
```

```python
# Subtotal

print("Subtotal:", subtotal)
```

```python
# Aplicar descuento

if subtotal > 200000:
    descuento = subtotal * 0.10
else:
    descuento = 0
```

```python
# Mostrar descuento

print("Descuento:", descuento)
```

```python
# Calcular total después del descuento

total_descuento = subtotal - descuento

print("Total después del descuento:", total_descuento)
```

```python
# Calcular el IVA

iva = total_descuento * 0.19
```

```python
# Calcular total final

total_final = total_descuento + iva
```

```python
# Muestra total a pagar

print(f"Total a pagar: ${total_final:,.0f}")
```

```python
# Ejercicio 3: Nómina con pandas
```

```python
# Creación de DataFrame

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "empleado": ["Cristian", "Jhon", "Lucia"],
    "area": ["Ventas", "RR.HH.", "TI"],
    "salario": [2500000, 3500000, 4500000]
})

print(df)
```

```python
# Creación nuevos empleados

nuevos = pd.DataFrame({
    "empleado": ["Juan", "Julian", "Martin"],
    "area": ["Ventas", "TI", "Ventas"],
    "salario": [2900000, 4800000, 3300000]
})

print(nuevos)
```

```python
# Unir Tablas

df = pd.concat([df, nuevos], ignore_index=True)

print(df)
```

```python
# Creación columna bono

df["bono"] = np.where(
    df["area"] == "Ventas",
    df["salario"] * 0.10,
    df["salario"] * 0.05
)

print(df)
```

```python
# Creación costo total

df["costo"] = df["salario"] + df["bono"]

print(df)
```

```python
# Calcular nómina por area

print(df.groupby("area")["costo"].sum())
```

```python
# Encontrar el salario mas alto

print(df.loc[df["salario"].idxmax(), "empleado"])
```

```python
# Respuesta

print("Costo total de nómina por área:")
print(df.groupby("area")["costo"].sum())

print("\nEmpleado con mayor salario:")
print(df.loc[df["salario"].idxmax(), "empleado"])
```
