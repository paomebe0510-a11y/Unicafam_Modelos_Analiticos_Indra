```python
# Ejercicio 3: Primer análisis de ventas
```

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
n = 200
```

```python
ventas = pd.DataFrame({
    "fecha": pd.date_range("2026-01-01", periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá", "Cartagena", "Medellin", "Barranquilla"], n),
    "canal": np.random.choice(["Web", "Tienda Fisica", "App"], n),
    "unidades": np.random.randint(1, 30, n),
    "precio": np.random.choice([35000, 60000, 100000], n),
})
```

```python
ventas["total"] = ventas["unidades"] * ventas["precio"]
```

```python
ventas.head()
```

```python
# ¿Cuál es el total vendido por canal (Tienda Fisica, Web, App)?

ventas.groupby("canal")["total"].sum()
```

```python
# ¿Cuál es el ticket promedio por ciudad?

ventas.groupby("ciudad")["total"].mean()
```

```python
# Los Días con mas ventas

ventas["dia_semana"] = ventas["fecha"].dt.day_name()

ventas.groupby("dia_semana")["total"].sum().idxmax()
```

 App es el canal con mayores ventas, seguida por la Web y por último la Tienda Física.Esto demuestra que las ventas digitales (App + Web) superan a la venta presencial, por lo que la empresa debe seguir fortaleciendo sus canales digitales, sin descuidar la tienda física.
