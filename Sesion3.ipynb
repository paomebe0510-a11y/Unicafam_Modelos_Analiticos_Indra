```python
# Mi presupuesto para cafe

presupuesto = 100000
precio_cafe = 15000
cafes_semana = 4

# ¿Cuánto gasto al mes?
gasto_mes = precio_cafe * cafes_semana * 4

# ¿Cuánto me sobra?
sobra = presupuesto - gasto_mes

# ¿Esto me alcanza alcanza? (True o False)
alcanza = presupuesto <= gasto_mes

print(gasto_mes, sobra, alcanza)
```
